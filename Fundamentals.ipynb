```python
# Create a list and print it
my_list = [1, 2, 3, 4, 5]
print(my_list)


# Access an item and print it
x = my_list[2]
print(x)


# Access last item with an index other than [4] and print it
y = my_list[-1]
print(y)

# Slice the list and print it
z = my_list[1:4]
print(z)

# Append an item and print it
my_list.append(6)
print(my_list)

# Remove an item and print it
my_list.remove(3)
print(my_list)

# Sort the list in descending order and print it
my_list.sort(reverse=True)
print(my_list)

# Get the length of the list and print it
length = len(my_list)
print(length)
```

```python
# Create a dictionary of the names and ages
people = {"Alice": "(20, female)", "Bob": "(30, male)", "Charlie": "(25, male)"}


# Add gender information to the dictionary
people["Alice"] = {"age": 20, "gender": "female"}
people["Bob"] = {"age": 25, "gender": "male"}
people["Charlie"] = {"age": 35, "gender": "male"}

print("\nUpdated dictionary with gender info:")
print(people)
```

```python
person = people["Alice"] #hier muss ich das erst einer neuen Variable zuweisen, sonst kann ich nicht auf die einzelnen Werte zugreifen (dic in dic); Python sucht immer nach den Keys und in meinem Dic sind das quasi Alice,... 
age = person["age"]
gender = person["gender"]

if age == 20:
    print(f'The age is 20 and the gender is {gender}.')
```

```python
#loopd and if-statements on the dictionary with Alice, Bob and Charlie

# Iterate through the dictionary with a loop printing the name and age
for x in people:
    print(f'Name: {x}, Age: {people[x]["age"]}')

# Iterate through the new dictionary printing both name, ages and gender
for y in people:
    print(f'Name: {y}, Age: {people[y]["age"]}, Gender: {people[y]["gender"]}')

# Iterate through the dictionary and print all males younger than 30
for z in people:
    if people[z]["gender"] == "male" and people[z]["age"] < 30:
        print(f'Name: {z}, Age: {people[z]["age"]}, Gender: {people[z]["gender"]}')
```
