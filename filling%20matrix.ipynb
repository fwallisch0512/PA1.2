```python
A77 = np.zeros((10,10))
A77[np.ix_([1,1,2,2],[1,2,1,2])] = 1
A77[np.ix_([5,5,6,6],[1,2,1,2])] = 1

A77[np.ix_([1,1,2,2],[5,6,5,6])] = 1
A77[np.ix_([5,5,6,6],[5,6,5,6])] = 1

A77[np.ix_([3,3,4,4],[3,3,4,4])] = 1
A77[np.ix_([3,3,4,4],[7,7,8,8])] = 1
plt.matshow(A77)
plt.show()
```

```python
A78 = np.zeros((5, 5))
A78[:, [0, 4]] = 1 #setzt linken und rechten Rand auf 1
A78[[1, 1, 2], [1, 3, 2]] = (
    1  #bedenke, dass Python 0 als erste Spalte/Zeile zählz
)



A79 =np.zeros((5, 5))
A79[:, [0, 4]] = 1 #setzt linken und rechten Rand auf 1
A79[[4, 4], [1, 3]] = 1

A80 = np.zeros((5, 5), dtype=int)
A80[[0, 4], :4] = 1
A80[1:4, [0, 4]] = 1

A81 = np.zeros((5, 5), dtype=int)
A81[::2, :] = 1
A81[:, 0] = 1


fig, axes = plt.subplots(2, 2)

axes[0, 0].matshow(A78)
axes[0, 1].matshow(A79)
axes[1, 0].matshow(A80)
axes[1, 1].matshow(A81)

plt.tight_layout()
plt.show()

```
